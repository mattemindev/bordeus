```python
from langchain_community.document_loaders import PDFPlumberLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Load the PDF using LangChain's PDFPlumber integration
pdf_path = "../knowledge/donnas/guide/Riciclabolario-generico-Teknoservice-NUOVO.pdf"

loader = PDFPlumberLoader(pdf_path)
documents = loader.load()

# 2. Configure the text splitter
# By removing spaces (" ") and characters ("") from the separators list,
# we force the splitter to keep entire lines together. This prevents
# it from chunking right through the middle of the items and bins.
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, chunk_overlap=100, separators=["\n\n", "\n"]
)

# 3. Split the document into chunks ready for embeddings
chunks = text_splitter.split_documents(documents)

# 4. Print results to verify
print(f"Successfully created {len(chunks)} chunks.\n")

if chunks:
    for i in range(len(chunks)):
        print(f"--- {i} Chunk Sample ---")
        print(chunks[i].page_content)

        print("\n--- Metadata ---")
        print(chunks[i].metadata)
```
